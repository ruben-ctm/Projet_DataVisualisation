```python
u, s = couts.usefulness_by_task(p26), couts.satisfaction_by_task(p26)
u.merge(s, on="task", suffixes=("_utilité", "_satisfaction"))[["task","net_utilité","net_satisfaction","n_utilité"]].round(3)
```

Utilité et satisfaction ne classent pas les tâches pareil : l'implémentation est jugée très utile, mais c'est aussi la tâche où la satisfaction baisse le plus souvent. À regarder avec les parts négatives, pas seulement le solde.
